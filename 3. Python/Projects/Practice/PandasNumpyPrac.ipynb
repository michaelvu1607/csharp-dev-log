```python
import numpy as np
import pandas as pd

# Set seed for reproducibility
np.random.seed(42)
num_sessions = 120

# 1. Generate Session Logs
player_ids = np.random.choice([f"P_{i:03d}" for i in range(1, 21)], size=num_sessions)
timestamps = pd.date_range(start="2026-08-01", periods=num_sessions, freq="2h")
duration_mins = np.random.uniform(10, 180, num_sessions)
damage_dealt = np.random.normal(5000, 1500, num_sessions).clip(500, 10000)
xp_gained = damage_dealt * np.random.uniform(0.1, 0.25, num_sessions)

# Introduce missing/corrupted values (dropped telemetry)
damage_dealt[::12] = np.nan

sessions_df = pd.DataFrame({
    'session_id': [f"S_{i:04d}" for i in range(1, num_sessions + 1)],
    'player_id': player_ids,
    'timestamp': timestamps,
    'duration_mins': np.round(duration_mins, 1),
    'damage_dealt': damage_dealt,
    'xp_gained': np.round(xp_gained, 0)
})
sessions_df.to_csv("player_sessions.csv", index=False)

# 2. Generate Store Purchases Log
num_purchases = 40
buyer_ids = np.random.choice(np.unique(player_ids), size=num_purchases)
item_types = np.random.choice(['Cosmetic', 'XP Boost', 'Battle Pass', 'Starter Pack'], size=num_purchases)
price_usd = np.random.choice([0.99, 4.99, 9.99, 19.99], size=num_purchases, p=[0.4, 0.3, 0.2, 0.1])

purchases_df = pd.DataFrame({
    'purchase_id': [f"TX_{i:04d}" for i in range(1, num_purchases + 1)],
    'player_id': buyer_ids,
    'item_type': item_types,
    'amount_usd': price_usd
})
purchases_df.to_csv("store_purchases.csv", index=False)

print("Setup Complete! 'player_sessions.csv' and 'store_purchases.csv' created.")
```

```python
ps = pd.read_csv("player_sessions.csv")
sp = pd.read_csv("store_purchases.csv")

ps = ps.fillna({"damage_dealt": 0})
print(ps)
```

```python
ps_dd_np = ps["damage_dealt"].to_numpy()
print(ps_dd_np)
ps_dm_np = ps["duration_mins"].to_numpy()
print(ps_dm_np)

dpm = ps_dd_np / ps_dm_np
print(dpm)
```

```python
ps["dpm"] = dpm
print(ps)
```

```python
sp_sum = sp["amount_usd"].sum()
print(sp_sum)
```

```python
merged_outer = pd.merge(ps, sp, on="player_id", how="outer")
print(merged_outer)
```

```python
merged_outer["Experienced"] = merged_outer["duration_mins"].apply(lambda x: "High Engagement" if x >= 60 else "Low Engagement")
print(merged_outer)
```

```python
merged_outer["Whales"] = merged_outer["amount_usd"].apply(lambda x: True if x >= 10 else False)
print(merged_outer)
```

```python
# find all paying players -> find avg dpm
paying_players = merged_outer[merged_outer["amount_usd"] > 1]
nonpaying_players = merged_outer[merged_outer["amount_usd"] < 1]
print(paying_players["dpm"].mean())
print(nonpaying_players["dpm"].mean())
```
