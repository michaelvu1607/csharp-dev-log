```python
import pandas as pd
import numpy as np

np.random.seed(42)

num_players = 100
player_ids = [f"P_{i:04d}" for i in range(1, num_players + 1)]

# Player Info
df_players = pd.DataFrame({
    "player_id": player_ids,
    "region": np.random.choice(["NA", "EU", "APAC"], size=num_players),
    "join_date": pd.date_range("2026-01-01", "2026-03-30", periods=num_players)
})

# Session Logs
session_player_ids = np.random.choice(player_ids, size=400)
durations = np.random.normal(loc=30, scale=15, size=400)
durations[np.random.choice(400, 20)] = np.nan  # missing values

df_sessions = pd.DataFrame({
    "session_id": [f"S_{i:05d}" for i in range(1, 400 + 1)],
    "player_id": session_player_ids,
    "duration_mins": durations,
    "score": np.random.randint(50, 500, size=400)
})

# Duplicate rows introduced
df_sessions = pd.concat([df_sessions, df_sessions.iloc[:10]], ignore_index=True)

# Save to CSV files
df_players.to_csv("df_players.csv", index=False)
df_sessions.to_csv("df_sessions.csv", index=False)

print("Saved df_players.csv and df_sessions.csv successfully!")
```

```python
df_sessions_clean = df_sessions.drop_duplicates(subset=None, keep="first")
print(df_sessions.to_string())
```

```python
median_dur = df_sessions_clean.loc[df_sessions_clean["duration_mins"] > 0, "duration_mins"].median()

df_sessions_clean["duration_mins"] = df_sessions_clean["duration_mins"].fillna(median_dur)
df_sessions_clean.loc[df_sessions_clean["duration_mins"] < 0, "duration_mins"] = median_dur
print(df_sessions_clean.to_string())
```

```python
score_per_min = df_sessions_clean["score"] / df_sessions_clean["duration_mins"]
df_sessions_clean["Score Per Min"] = score_per_min
print(df_sessions_clean)
```
